```python
import sys
import importlib.util
import subprocess

required = ["pandas", "sqlalchemy", "psycopg2"]
missing = [pkg for pkg in required if importlib.util.find_spec(pkg.replace('-', '_')) is None]

if not missing:
    print("Entorno listo. Todas las dependencias requeridas están instaladas.")
else:
    print(f"Faltan dependencias: {missing}. Instalando requirements.txt...")
    subprocess.check_call([sys.executable, '-m', 'pip', 'install', '-r', 'requirements.txt'])
    print("Instalación completada. Reinicia el kernel si es necesario.")
```
